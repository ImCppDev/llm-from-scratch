```python
def grad_f3(x, y):
    """Analytical gradient of f3 w.r.t. (x, y).

    TODO: work out df3/dx and df3/dy on paper, then return them as
    the tuple (df3/dx, df3/dy).
    """
    return (
        2 * x * y,
        x ** 2 + 3 * (y ** 2)
    )
```

```python
grad = grad_f3(2,3)
```

```python
print(grad)
```

```python
def f3(x, y):
    """f3(x, y) = x**2 * y + y**3"""
    return x ** 2 * y + y ** 3
```

```python
x0,y0 = 2,3
x1,y1=x0 + 0.01 * grad[0],y0 + 0.01 * grad[1]
f0 = f3(x0,y0)
f1 = f3(x1,y1)
x0,y0,x1,y1,f0,f1
```

```python
x0,y0 = 2,3
x2,y2=x0 - 0.01 * grad[0],y0 - 0.01 * grad[1]
f2 = f3(x2,y2)
x0,y0,x1,y1,x2,y2,f0,f1,f2
```
